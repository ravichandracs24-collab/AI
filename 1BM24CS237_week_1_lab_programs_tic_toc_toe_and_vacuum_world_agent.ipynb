```python
# Tic-Tac-Toe using AI

board = [" " for _ in range(9)]


def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


def board_full():
    return " " not in board


def computer_move():
    # Try to win
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"

            if check_winner("O"):
                return

            board[i] = " "

    # Block player
    for i in range(9):
        if board[i] == " ":
            board[i] = "X"

            if check_winner("X"):
                board[i] = "O"
                return

            board[i] = " "

    # Take center
    if board[4] == " ":
        board[4] = "O"
        return

    # Take a corner
    for i in [0, 2, 6, 8]:
        if board[i] == " ":
            board[i] = "O"
            return

    # Take any remaining space
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            return


print("TIC-TAC-TOE")
print("You = X")
print("Computer = O")

while True:

    display_board()

    position = int(input("Enter position (1-9): ")) - 1

    if position < 0 or position > 8 or board[position] != " ":
        print("Invalid position!")
        continue

    board[position] = "X"

    if check_winner("X"):
        display_board()
        print("You Win!")
        break

    if board_full():
        display_board()
        print("Draw!")
        break

    computer_move()

    if check_winner("O"):
        display_board()
        print("Computer Wins!")
        break

    if board_full():
        display_board()
        print("Draw!")
        break
```

```python
# Vacuum Cleaner World using Agent

room = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"


def vacuum_agent(location, room):

    print("\nAgent is in Room", location)
    print("Room A:", room["A"])
    print("Room B:", room["B"])

    # If current room is dirty
    if room[location] == "Dirty":

        print("Action: SUCK")
        room[location] = "Clean"

    # If current room is clean
    else:

        if location == "A":
            print("Action: MOVE RIGHT")
            location = "B"

        else:
            print("Action: MOVE LEFT")
            location = "A"

    return location


print("VACUUM CLEANER WORLD")
print("=====================")

print("\nInitial State:")
print("Room A:", room["A"])
print("Room B:", room["B"])
print("Agent Location:", location)

# Run agent
for i in range(4):

    location = vacuum_agent(location, room)

print("\nFinal State:")
print("Room A:", room["A"])
print("Room B:", room["B"])
print("Agent Location:", location)
```

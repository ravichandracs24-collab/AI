```python

def astar(graph, heuristic, start, goal):
    open_list = [(heuristic[start], 0, start)]
    visited = set()
    parent = {}

    while open_list:
        open_list.sort()

        f, g, current = open_list.pop(0)

        if current == goal:
            path = []
            while current in parent:
                path.append(current)
                current = parent[current]
            path.append(start)
            path.reverse()

            print("Path:", path)
            print("Cost:", g)
            return

        if current in visited:
            continue

        visited.add(current)

        for neighbor, cost in graph[current].items():
            if neighbor not in visited:
                new_g = g + cost
                new_f = new_g + heuristic[neighbor]

                parent[neighbor] = current
                open_list.append((new_f, new_g, neighbor))

    print("Goal not found")


graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'E': 1},
    'D': {'G': 5},
    'E': {'G': 3},
    'G': {}
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 4,
    'D': 4,
    'E': 2,
    'G': 0
}

astar(graph, heuristic, 'A', 'G')
```

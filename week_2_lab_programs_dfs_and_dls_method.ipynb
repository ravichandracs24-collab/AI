```python
# 8-Puzzle using DFS

GOAL = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)


def display(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


def get_neighbors(state):

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    neighbors = []

    directions = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in directions:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(
                (tuple(new_state), move)
            )

    return neighbors


def DFS(state, path, visited):

    # Goal test
    if state == GOAL:
        return path

    for new_state, move in get_neighbors(state):

        if new_state not in visited:

            visited.add(new_state)

            result = DFS(
                new_state,
                path + [(move, new_state)],
                visited
            )

            if result is not None:
                return result

    return None


# Initial state
start = (
    1, 2, 3,
    4, 5, 6,
    0, 7, 8
)


# Run DFS
solution = DFS(start, [], {start})


# Display solution
print("DFS SOLUTION")
print("============")

print("\nInitial State:")
display(start)

if solution:

    for i, (move, state) in enumerate(solution, 1):

        print("Step", i, ":", move)
        display(state)

else:
    print("Goal not found")
```

```python
# 8-Puzzle using Depth-Limited Search (DLS)

GOAL = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

nodes = 0


def display(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def get_neighbors(state):
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    neighbors = []

    directions = [
        (-1, 0, "Up"),
        (1, 0, "Down"),
        (0, -1, "Left"),
        (0, 1, "Right")
    ]

    for dr, dc, move in directions:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append((tuple(new_state), move))

    return neighbors


# Depth-Limited Search
def DLS(state, depth, path, visited):

    global nodes
    nodes += 1

    # Goal test
    if state == GOAL:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    for new_state, move in get_neighbors(state):

        if new_state not in visited:

            visited.add(new_state)

            result = DLS(
                new_state,
                depth - 1,
                path + [(move, new_state)],
                visited
            )

            if result is not None:
                return result

            visited.remove(new_state)

    return None


# Initial state
start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)


# Depth limit = 5
LIMIT = 5

solution = DLS(
    start,
    LIMIT,
    [],
    {start}
)


print("DLS SOLUTION")
print("============")

print("\nInitial State:")
display(start)

if solution:

    for i, (move, state) in enumerate(solution, 1):
        print("Step", i, ":", move)
        display(state)

    print("Number of moves:", len(solution))

else:
    print("Goal not found within depth limit.")

print("Depth Limit:", LIMIT)
print("Nodes Explored:", nodes)
print("Time Complexity: O(3^5) = O(243)")
```

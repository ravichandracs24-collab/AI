```python

def unify(x, y, subs=None):
    if subs is None:
        subs = {}

    if x == y:
        return subs

    if isinstance(x, str) and x.islower():
        return unify_var(x, y, subs)

    if isinstance(y, str) and y.islower():
        return unify_var(y, x, subs)

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y):
            return None

        for a, b in zip(x, y):
            subs = unify(a, b, subs)
            if subs is None:
                return None

        return subs

    return None


def unify_var(var, value, subs):
    if var in subs:
        return unify(subs[var], value, subs)

    if var == value:
        return subs

    subs[var] = value
    return subs


# Main program
expr1 = ("Knows", "John", "x")
expr2 = ("Knows", "y", "Mary")

result = unify(expr1, expr2)

if result is not None:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failed")
```

```python

def forward_chaining(facts, rules, query):

    new_facts = list(facts)

    while True:
        new = []

        for rule in rules:
            premise1, premise2, conclusion = rule

            for f1 in new_facts:
                for f2 in new_facts:

                    if (f1[0] == premise1[0] and
                        f2[0] == premise2[0] and
                        f1[2] == f2[1]):

                        x = f1[1]
                        z = f2[2]

                        result = (conclusion[0], x, z)

                        if result not in new_facts and result not in new:
                            new.append(result)

                        if result == query:
                            return True

        if not new:
            return False

        new_facts.extend(new)


# Knowledge Base - Facts
facts = [
    ("Parent", "John", "Mary"),
    ("Parent", "Mary", "Sam")
]

# Rule
rules = [
    (
        ("Parent", "X", "Y"),
        ("Parent", "Y", "Z"),
        ("Grandparent", "X", "Z")
    )
]

# Query
query = ("Grandparent", "John", "Sam")

# Execute
if forward_chaining(facts, rules, query):
    print("Query Proved")
else:
    print("Query Not Proved")
```

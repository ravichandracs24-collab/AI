```python
import heapq

start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Number of misplaced tiles
def heuristic(state):
    h = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            h += 1

    return h


# Generate possible states
def neighbors(state):
    result = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:

        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:

            new_blank = r * 3 + c

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            result.append(tuple(new_state))

    return result


# A* algorithm
def a_star(start):

    pq = []

    g = 0
    h = heuristic(start)
    f = g + h

    heapq.heappush(pq, (f, g, start))

    cost = {start: 0}
    parent = {start: None}

    while pq:

        f, g, current = heapq.heappop(pq)

        if current == goal:
            return parent, current

        for next_state in neighbors(current):

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g

                h = heuristic(next_state)
                f = new_g + h

                parent[next_state] = current

                heapq.heappush(
                    pq, (f, new_g, next_state)
                )

    return None, None


# Print solution
def print_solution(parent, state):

    path = []

    while state is not None:
        path.append(state)
        state = parent[state]

    path.reverse()

    for step, state in enumerate(path):

        print("Step", step)

        for i in range(0, 9, 3):
            print(state[i:i+3])

        print()


parent, final = a_star(start)

if final:
    print_solution(parent, final)
else:
    print("No solution")
```

```python
import heapq

# Goal state
GOAL = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

# Initial state
START = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)


# Manhattan Distance
def heuristic(state):
    distance = 0

    for i in range(9):

        # Do not count blank tile
        if state[i] == 0:
            continue

        # Current position
        current_row = i // 3
        current_col = i % 3

        # Goal position
        goal_index = GOAL.index(state[i])
        goal_row = goal_index // 3
        goal_col = goal_index % 3

        # Manhattan distance
        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# Generate possible moves
def get_neighbors(state):

    neighbors = []

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_position = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with tile
            new_state[zero], new_state[new_position] = \
                new_state[new_position], new_state[zero]

            neighbors.append((tuple(new_state), move))

    return neighbors


# A* Search
def a_star(start):

    priority_queue = []

    g = 0
    h = heuristic(start)
    f = g + h

    heapq.heappush(
        priority_queue,
        (f, g, start, [])
    )

    visited = set()

    while priority_queue:

        f, g, state, path = heapq.heappop(priority_queue)

        if state in visited:
            continue

        visited.add(state)

        # Goal reached
        if state == GOAL:
            return path + [(state, "GOAL")]

        for neighbor, move in get_neighbors(state):

            if neighbor not in visited:

                new_g = g + 1
                new_h = heuristic(neighbor)
                new_f = new_g + new_h

                heapq.heappush(
                    priority_queue,
                    (
                        new_f,
                        new_g,
                        neighbor,
                        path + [(state, move)]
                    )
                )

    return None


# Print board
def print_board(state):

    for i in range(0, 9, 3):
        print(
            state[i],
            state[i + 1],
            state[i + 2]
        )


# Run A*
solution = a_star(START)

print("CASE 2 - A* SEARCH")
print("Heuristic = Manhattan Distance")
print()

if solution:

    for step, (state, move) in enumerate(solution):

        print("Step:", step)

        if step > 0:
            print("Move:", move)

        print_board(state)

        g = step
        h = heuristic(state)
        f = g + h

        print("g(n) =", g)
        print("h(n) =", h)
        print("f(n) =", f)

        print("-------------------")

else:
    print("No solution found.")
```

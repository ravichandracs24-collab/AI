```python

def alpha_beta(depth, node, maximizing, values, alpha, beta):
    # Base condition
    if depth == 3:
        return values[node]

    if maximizing:
        best = float('-inf')

        for i in range(2):
            val = alpha_beta(depth + 1, node * 2 + i,
                             False, values, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            val = alpha_beta(depth + 1, node * 2 + i,
                             True, values, alpha, beta)
            best = min(best, val)
            beta = min(beta, best)

            if beta <= alpha:
                break

        return best


# Leaf node values
values = [3, 5, 6, 9, 1, 2, 0, -1]

result = alpha_beta(0, 0, True, values,
                    float('-inf'), float('inf'))

print("Optimal value:", result)
```

```python

import random
import math

def simulated_annealing():

    current = 10
    best = current

    temperature = 100
    cooling_rate = 0.95

    while temperature > 0.1:

        # Generate a neighboring solution
        new = current + random.choice([-1, 1])

        if new < -10 or new > 10:
            temperature *= cooling_rate
            continue

        # Calculate cost function
        current_cost = current ** 2
        new_cost = new ** 2

        # Accept better solution
        if new_cost < current_cost:
            current = new

        # Accept worse solution with probability
        else:
            probability = math.exp(
                (current_cost - new_cost) / temperature
            )

            if random.random() < probability:
                current = new

        # Update best solution
        if current ** 2 < best ** 2:
            best = current

        # Reduce temperature
        temperature *= cooling_rate

    print("Best Solution:", best)
    print("Minimum Cost:", best ** 2)


# Execute algorithm
random.seed(42)
simulated_annealing()
```

```python

from itertools import product

# Knowledge Base
# P = It is raining
# Q = Roads are wet
# R = Traffic is slow

symbols = ['P', 'Q', 'R']

# KB: P -> Q and Q -> R
def knowledge_base(model):
    P, Q, R = model

    return ((not P or Q) and
            (not Q or R))

# Query: P -> R
def query(model):
    P, Q, R = model
    return not P or R

# Check entailment
def check_entailment():
    entails = True

    for model in product([False, True], repeat=3):
        if knowledge_base(model):
            print("Model:", model)

            if not query(model):
                entails = False

    if entails:
        print("\nKB entails the Query")
    else:
        print("\nKB does not entail the Query")


check_entailment()
```
